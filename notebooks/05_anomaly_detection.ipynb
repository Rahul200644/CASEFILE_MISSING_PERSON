# Module 5: Movement Anomaly Detection
Detecting unusual movements using Isolation Forest, LOF, and One-Class SVM.

```python
import sys
sys.path.append('../')
import pandas as pd
import joblib
from src.anomaly_detection import detect_anomalies

df = pd.read_csv('../data/processed/gps_trajectories_processed.csv')
iso_model = joblib.load('../models/anomaly_model.pkl')
scaler = joblib.load('../models/anomaly_scaler.pkl')

features = ['calculated_speed_kmh', 'step_distance_km', 'hour', 'is_weekend']
X_scaled = scaler.transform(df[features].fillna(0.0))
df['is_anomaly'] = iso_model.predict(X_scaled) == -1
print('Total Anomalies:', df['is_anomaly'].sum())
```
