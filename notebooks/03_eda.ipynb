# Module 3 & 4: Exploratory Data Analysis (EDA)
Analyzing temporal patterns, speed distributions, and geographical coordinates.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/processed/gps_trajectories_processed.csv')

plt.figure(figsize=(10, 4))
sns.histplot(df['calculated_speed_kmh'], bins=30, kde=True, color='purple')
plt.title('Speed Distribution (km/h)')
plt.show()
```
